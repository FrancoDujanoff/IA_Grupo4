## Ejercicios Teóricos

1. ¿Qué es una inferencia?

Es el proceso mediante el cual un agente llega a nuevas conclusiones a partir de conocimiento que ya tiene. Es decir que a partir de hechos o reglas conocidas, deduce o interfiere informacion nueva

Por ejemplo, si un agente sabe que "La oficina esta sucia, entonces hay que limpiarla"

Y tambien sabe que "La oficina esta sucia"

Entonces infiere que hay que limpiar la oficina

2. ¿Cómo se verifica que un modelo se infiere de la base de conocimientos?

Para verificar que una sentencia se infiere de la base de conocimiento, se revisa si es verdadera en todos los modelos donde la base de de conocimientos tambien lo es. Osea que queremos saber si una conclusion es logicamente consecuencia de lo que ya sabemos (la base de conocimiento).

3. Observe la siguiente base de conocimiento:

$R1: b ∧ c → a$

$R2: d ∧ e → b$

$R3: g ∧ e → b$

$R4: e → c$

$R5: d$

$R6: e$

$R7: a ∧ g → f$

        3.1 ¿Cómo se puede probar que $a = True$ a través del encadenamiento hacia adelante? Este método solamente usa reglas ya incorporadas a la base de conocimiento para inferir la hipótesis, ¿qué propiedad debe tener el algoritmo para asegurar que esta inferencia sea posible?

        3.2 ¿Cómo se puede probar que $a = True$ a través del encadenamiento hacia atrás? Este método asigna un valor de verdad a la hipótesis y deriva las sentencias de la base de conocimiento, ¿qué propiedad debe tener el algoritmo para asegurar que esta derivación sea posible?

        3.3 Exprese la base de conocimiento en su Forma Normal Conjuntiva. A continuación, demuestre por contradicción que $a = True$.

3.1 -> Este metodo parte de los hechos conocidos (R5 y R6) y se van aplicando reglas para inferir en nuevas verdades hasta llegar a la meta. Entonces por R5 y R6 sabemos que d y e son verdaderos. Por R2, sabemos que b sera verdadero. Por R4, sabemos que c sera verdadero. Por R1, llegamos entonces a que a es verdadero, cumpliendo asi la meta.
La propiedad requerida para que el algoritmo funcione correctamente es que "debe ser completo"

3.2 ->En este ahora partimos de la meta, osea que a es verdadero y retrocedemos a ver que hechos necesitamos para que esto sea asi. Por R1 para que a sea verdadero, necesitamos que b y c sea verdadero, por R2 para que b sea verdadero, d y e tienen que ser verdadero, por R4 si e es verdadero, entonces c sera verdadero. Ahora teniendo b y c verdadero por R1 se hace verdadero a, asi que se cumple. 
La propiedad que se necesita para que este metodo funcione bien es que "el algorimo debe ser solido"

3.3 -> Sabiendo que la base de conocimiento es:
R1: b ∧ c→a
R2: d ∧ e→b
R3: g ∧ e→b
R4: e→c
R5: d
R6: e
R7: a∧g→f

1. Pasar todo a la forma normal conjuntiva (FNC).La FNC es una conjunción de cláusulas (cada una disyunción de literales).
La BC en FNC es:
R1. ¬b ∨¬c ∨ a  (es la equivalencia de b ∧ c→a)
R2. ¬d ∨ ¬e ∨ b (equivalencia de  d ∧ e→b)
R3. ¬g ∨ ¬e ∨ b (equivalencia de  g ∧ e→b)
R4. ¬e ∨ c      (equivalencia de  e→c)
R5. d           (ya es un literal positivo)
R6. e           (ya es un literal positivo)
R7. ¬a ∨ ¬g ∨ f (equivalencia de a∧g→f)

2. Demostración por contradicción de que a=True
Para probar que a es verdadero:
Asumimos lo contrario: ¬a.
Si la BC ∧ ¬a lleva a contradicción, entonces a es verdadero.

Paso 1: Partimos de los hechos (Usamos los nuevos R no los anteriores) De R5 y R6 tenemos: d verdadero; e verdadero
Paso 2: Usamos R2 sustituyendo d y e verdaderos: (F=falso)  ¬d ∨ ¬e ∨ b = F ∨ F ∨ b = b . Entonces → b es verdadero
Paso 3: Usamos R4 Con e verdadero: ¬e ∨ c =F ∨ c = c . Entonces → c es verdadero
Paso 4: Usamos R1 Con b y c verdaderos: ¬b ∨ ¬c ∨ a = F ∨ F ∨ a = a . Entonces → a es verdadero

3. Contradicción
Primero asumimos que "¬a" era verdadero como suposición pero llegamos a que "a" es verdadero por derivación lógica generando una contradicción con la suposición "¬a".
Por lo tanto por reducción al absurdo, la BC ∧ ¬a es inconsistente. Por lo tanto:
a=Truea
