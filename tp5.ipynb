1. ¿Cuáles son los tres axiomas de Kolmogorov?

-No negatividad: Para cualquier evento A, la probabilidad es siempre un numero mayor o igual a cero.

-Normalizacion: La probabilidad del suceso seguro es 1

-Aditividad: Si A y B son eventos mutuamente excluyentes, la probabilidad de su union es la suma de sus probabilidades, y si si tienen una realcion, union, la probabilidad de que ocurran ambas es la suma de sus probabilidades menos la probabilidad de su union.

Estos axiomas garantizan que la probabilidad sea una medida coherente y permiten derivar todas las demas reglas.
