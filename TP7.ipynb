1. ¿Qué representa una función de utilidad?

La funcion de utilidad es una representacion numerica de nuestras preferencias. No asigna un valor numero a la preferencia simplemente dicha, sino que simplemente indican orden y magnitud de preferencia. Se basa en en axiomas como Ordenacion, Transitividad, Continuidad. Simplemente traduce las preferencias cualitativas de un agente en una escala cuantitativa, lo que permite aplicar matematicamente el principio de maximizacion de la utilidad (MUE), asi el agente selecciona la accion que maximiza la utilidad promedio esperada, considerando tanto los posibles resultados como sus probabilidades

&emsp;&emsp;2.1 ¿Qué color prefieren?

```python
import requests
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt

# URLs directas de Google Drive
url1 = "https://drive.google.com/uc?export=view&id=1aofwvS7bUVmKS6MlTcpL2ZgVa9JuGJex"
url2 = "https://drive.google.com/uc?export=view&id=1-htp7twx9y5-GpzMUXl9hk-X3kV-BVz6"

# Función para descargar y redimensionar imagen
def load_and_resize(url, size=(10, 10)):
    response = requests.get(url)
    img = Image.open(BytesIO(response.content))
    img = img.resize(size, Image.Resampling.LANCZOS)  # redimensionar a 150x150
    return img

# Cargar imágenes
img1 = load_and_resize(url1)
img2 = load_and_resize(url2)

# Crear figura con 2 columnas
fig, axes = plt.subplots(1, 2, figsize=(5, 5))

# Mostrar primera imagen
axes[0].imshow(img1)
axes[0].axis('off')
axes[0].set_title("Color 1")

# Mostrar segunda imagen
axes[1].imshow(img2)
axes[1].axis('off')
axes[1].set_title("Color 2")

plt.tight_layout()
plt.show()
```

&emsp;&emsp;2.2 Entre sacar un 10 o un 7 en un parcial, ¿qué prefieren? ¿Y entre el 7 y desaprobar?

Se relaciona con el axioma de "Transtividad" que establece que entre tres estados, si un agente prefiere A frente a B y a B frente a C, entonces prefiere A frente a C. Aplicado a este caso se prefiere sacarse un 7 antes que desaprobar, y sacarse un 10 antes un 7, entonces se va a preferir aprobar antes que sacarse un 10.


&emsp;&emsp;2.3 Imagínense que hacen una apuesta de $100 jugando al cara o cruz, si pudieran trucar la moneda para que ustedes ganaran un porcentaje *p* de las veces, ¿en qué valor configurarían *p*?

Se refiere al axioma "Continuidad" el cual establece que si existe algun estado B entre A y C en la relacion de preferencia, entonces existe un valor de probabilidad p para que al agente racional le es indiferente considerar como seguro al estado B, y la literal establece el estado A con probabilidad p y al estado C con probabilidad 1-p. Es decir que configurar un p en este caso es justamente una traduccion practica del axioma, donde existe un valor intermedio de probabilidad que equilibra la preferencia entre ganar y perder, por lo que por ejemplo si tenemos una probabilidad p de ganar del 70% que es el estado A, tendriamos una probabilidad del 30% de perder, que seria el estado C.


&emsp;&emsp;2.4 En una feria hay dos juegos: uno en el que el premio son $100 y otro en el que el premio es una piedra, ambos con una probabilidad *p* de ganar al juego. ¿A qué juego prefieren jugar?

Se relaciona con el axioma de "Sustitucion", dado que el agente es indiferente entre estas dos loterias entonces el agente va a ser indiferente en dos loterias mas complejas en las que aparecen estas. Entonces se preferira el primero debido a que su utilidad esperada es mucho mayor que la segunda.


&emsp;&emsp;2.5 En un casino hay dos juegos y en ambos el premio son $1000. El primer juego se gana cuando se tira una moneda y sale cara, y el segundo juego se gana cuando sale 6 en un dado. ¿A qué juego prefieren jugar?

Se relaciona con el axioma de "Monotonicidad" dado que hay dos loterias que presentan dos resultados posibles e iguales (osea en ambas se gana 1000), se debe preferir la loteria que presente una mayor probabilidad para que el resultado preferido ocurra, o sea la de tirar la moneda.

5. Una empresa minera quiere enviar a un robot recolector de materiales a visitar un planeta. El costo del viaje asciende a \$20.000.000. Tiene que decidir a qué planeta lo enviará de entre 5 opciones, sabiendo que uno de ellos contiene oro por un valor de \$100.000.000 y los no poseen nada de valor.<br>
La Comisión Nacional del Espacio ofrece los resultados de un estudio sobre el planeta X en el que indica si hay o no oro; y vende dicha información por \$25.000.000. ¿Nos beneficia comprar dicha información?

Fundamento Teórico: Valor de la Información Perfecta (VIP)
Para resolver este problema, utilizamos el concepto de Valor de la Información Perfecta (VIP). Como vimos en los apuntes de "Toma de Decisiones", el VIP se define como la diferencia entre el valor esperado de la mejor acción.La idea es calcular el "valor máximo" que estaríamos dispuestos a pagar por tener información certera antes de tomar una decisión, de igual forma quedará a criterio del agente tomar o no la decisión de tomar o rechazar la información.
Después de tener la información y el valor esperado de la mejor acción antes de tenerla. En esencia, nos dice cuánto vale para nosotros tener información 100% certera.

Análisis del Problema
Paso 1: Calcular el Valor Monetario Esperado (VME) sin la información
Si no compramos el estudio, la decisión es enviar el robot a un planeta al azar.

Probabilidad de éxito (encontrar oro): p(exito)= 1/5
​Probabilidad de fracaso (no encontrar oro): p(fracaso)= 4/5
​Ganancia neta en caso de éxito: $100.000.000−$20.000.000=$80.000.000
Ganancia neta en caso de fracaso: −$20.000.000

La fórmula del Valor Monetario Esperado (VME) es:
VME(sin_info)=[p(exito)×Ganancia neta]+[p(fracaso)×Perdida neta]

Reemplazando los valores:
VME(sin_info)=[(1/5)×($80.000.000)]+[(4/5)×(−$20.000.000)]
VME(sin_info) =$16.000.000−$16.000.000=$0

Sin información, el beneficio esperado de la misión es de $0.

Paso 2: Calcular el Valor Monetario Esperado (VME) con la información perfecta
Ahora, calculamos el valor esperado si tuviéramos la información del estudio.

Escenario A: El estudio confirma que SI hay oro en el planeta X.
Probabilidad de este escenario: P(A)= 1/5
​Decisión: Enviar el robot al planeta X.
Resultado: Ganancia segura de $80.000.000.

Escenario B: El estudio confirma que NO hay oro en el planeta X.
Probabilidad de este escenario: P(B)=4/5
​Decisión: Sabemos que el oro está en uno de los 4 planetas restantes. La mejor opción es explorar uno de ellos.

VME en este escenario:
Nueva probabilidad de éxito: 1/4
​VME(B)=(1/4)×($80.000.000)+(3/4)×(−$20.000.000)=$20.000.000−$15.000.000=$5.000.000

El VME total con información es el promedio ponderado de los resultados de cada escenario:
VME(con_info)=P(A)×Resultado(A)+P(B)×VME(B)
VME(con_info)=(1/5)×$80.000.000+(4/5)×$5.000.000
VME(con_info)=$16.000.000+$4.000.000=$20.000.000
Con información perfecta, el beneficio esperado de la misión sube a $20.000.000.

Paso 3: Calcular el Valor de la Información Perfecta (VIP)
Ahora podemos calcular el valor real de la información.
VIP=VME(con_info)−VME(sin_info)
​VIP=$20.000.000−$0=$20.000.000

Conclusión
Finalmente, comparamos el valor de la información con su precio de venta:
Valor de la Información (VIP): $20.000.000
Costo del Estudio: $25.000.000
Dado que Costo > VIP, la empresa estaría perdiendo $5.000.000 si decide comprar el estudio. Por lo tanto, la decisión racional es no comprar la información pero teniendo en cuenta que en estos casos la decisión es por utilidad y no por racionalidad queda a criterio de empresa.
